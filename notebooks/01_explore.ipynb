**Row 0: Vektis rest category** (source: Vektis leaflet, p.4)
- Vektis puts groups with fewer than 10 people (for privacy) and people with unknown sex, age or municipality into one row with empty labels.
- This explains why the file has 12,992 rows instead of 12,996.
- This row can't be assigned to a municipality, so it is excluded from the municipality comparison during cleaning. The raw file stays unchanged.


**Leaflet notes** (Vektis bijsluiter 2011-2024)

1. **Insured-years, not people (p.5-6).** 
   
   why do you divide costs by insured-years and not by the number of people? 
   
   Some people are insured for only part of the year (born, died, moved). Dividing costs by insured-years gives a fair cost per full year of cover, so regions with many movers are not made to look cheaper.

2. **Municipality borders (p.5).** 

   the 2024 file uses the municipality borders of which year? And why could that cause trouble later, when you join it with the CBS health data?

   The 2024 file uses the **2026** municipality borders.Before joining with CBS, check which border year CBS uses: municipalities that merged between those years will not match by name. These go into `name_fixes.csv`.

3. **Age (p.5).** 
   how wide is each age band, what's the oldest band, and on which date is age measured?

   Five-year age bands at municipality level, with 90+ as the top band. Age is measured on 31 December 2024.

4. **Rest category (p.4).** 

   See the note on row 0 above.

5. **Care types and completeness (p.7-10).**
   
   look up what mdz and gzsp mean. Then read 4.3 "uitloopkosten": are the 2024 costs complete, or could some still be missing?

   - `mdz` = multidisciplinary GP care: team programmes for chronic conditions, plus GP performance and innovation payments.
   - `gzsp` = medical care for specific patient groups (category since 2020). The leaflet gives no further detail; to confirm.
   - Costs include claims processed up to 8 quarters (2 years) after 1 Jan 2024. Most care is then complete, but **specialist mental health care is only about 90% complete** and cross-border care is incomplete.
   - Hospital costs are claimed prices; ceiling agreements can lower them afterwards.
